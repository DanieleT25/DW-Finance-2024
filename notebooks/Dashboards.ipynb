## Preprocessing

```python
lst_comparison = ['symbol', 'company_name', 'sector', 'industry', 'country', 'market_cap']

lst_attribute = ['net_income', 'total_revenue', 'return_on_assets', 'return_on_equity', 'profit_margins',
                 'pe_trailing', 'pe_forward', 'earnings_growth', 'price_to_sales', 'price_to_book',
                 'revenue_growth', 'debt_to_equity', 'dividend_yield', 'payout_ratio',
                 'free_cashflow']
```

```python
df = data[lst_comparison + lst_attribute].copy()
```

```python
df['net_profit_margin']     = df['net_income'] / df['total_revenue']
df['peg_ratio']             = df['pe_forward'] / df['earnings_growth']

df = df.rename(columns={
    'return_on_assets': 'roa',
    'return_on_equity': 'roe'
})
```

```python
df.describe()
```

There are some strange values

```python
df_cleaned = df.copy()

df_cleaned = df_cleaned[df_cleaned['net_income'] <= 1e12]
df_cleaned = df_cleaned[df_cleaned['free_cashflow'] <= 1e12]
df_cleaned = df_cleaned[np.isfinite(df_cleaned['pe_trailing'])]
df_cleaned = df_cleaned[np.isfinite(df_cleaned['price_to_sales'])]

df_cleaned = df_cleaned[np.isfinite(df_cleaned['net_profit_margin'])]
df_cleaned = df_cleaned[np.isfinite(df_cleaned['pe_forward'])]
df_cleaned = df_cleaned[np.isfinite(df_cleaned['peg_ratio'])]

(len(df)-len(df_cleaned))/len(df)
```

Too much data has been removed, so it is better to set it NaN.

In plots and comparisons, NaN values are automatically ignored. So: 
* We retain the maximum amount of data for all metrics
* We avoid removing relevant companies with only partial missing values

```python
df_cleaned = df.copy()

df_cleaned.loc[df_cleaned['net_income'] > 1e12, 'net_income'] = np.nan
df_cleaned.loc[df_cleaned['free_cashflow'] > 1e12, 'free_cashflow'] = np.nan
df_cleaned.loc[df_cleaned['price_to_sales'] <= 0, 'price_to_sales'] = np.nan
df_cleaned.loc[~np.isfinite(df_cleaned['price_to_sales']), 'price_to_sales'] = np.nan
df_cleaned.loc[~np.isfinite(df_cleaned['pe_trailing']), 'pe_trailing'] = np.nan

df_cleaned['net_profit_margin'] = df_cleaned['net_profit_margin'].where(np.isfinite(df_cleaned['net_profit_margin']), np.nan)
df_cleaned['pe_forward'] = df_cleaned['pe_forward'].where(np.isfinite(df_cleaned['pe_forward']), np.nan)
df_cleaned['peg_ratio'] = df_cleaned['peg_ratio'].where(np.isfinite(df_cleaned['peg_ratio']), np.nan) 
```

```python
q1 = df_cleaned['payout_ratio'].quantile(0.25)
q3 = df_cleaned['payout_ratio'].quantile(0.75)
iqr = q3 - q1

upper_bound = q3 + 1.5 * iqr
df_cleaned.loc[df_cleaned['payout_ratio'] > upper_bound, 'payout_ratio'] = np.nan
```

Let us improve the dataframe's structure

```python
df_base = df_cleaned[lst_comparison].copy()

df_base.columns = pd.MultiIndex.from_tuples([('Info', col) for col in df_base.columns])
```

```python
groups = {
    'Profitability': {
        'net_profit_margin': df_cleaned['net_profit_margin'],
        'roa': df_cleaned['roa'],
        'roe': df_cleaned['roe'],
        'profit_margins': df_cleaned['profit_margins']
    },
    'Valuation': {
        'pe_trailing': df_cleaned['pe_trailing'],
        'pe_forward': df_cleaned['pe_forward'],
        'peg_ratio': df_cleaned['peg_ratio'],
        'price_to_sales': df_cleaned['price_to_sales'],
        'price_to_book': df_cleaned['price_to_book']
    },
    'Efficiency': {
        'revenue_growth': df_cleaned['revenue_growth'],
        'earnings_growth': df_cleaned['earnings_growth']
    },
    'Financial Strength': {
        'debt_to_equity': df_cleaned['debt_to_equity'],
        'dividend_yield': df_cleaned['dividend_yield'],
        'payout_ratio': df_cleaned['payout_ratio'],
        'free_cashflow': df_cleaned['free_cashflow']
    }
}
```

```python
metric_data = {}
for group, metrics in groups.items():
    for metric_name, series in metrics.items():
        metric_data[(group, metric_name)] = series

df_metrics = pd.DataFrame(metric_data)
df_metrics.columns = pd.MultiIndex.from_tuples(df_metrics.columns)
```

```python
df_comp = pd.concat([df_base, df_metrics], axis=1)
```

```python
pd.set_option('display.max_columns', None)
df_comp
```

## Peer Comparison

```python
companySymbol = 'ATR'
```

```python
company = df_comp[df_comp['Info', 'symbol'] == companySymbol]
company
```

**Which metrics were selected?**

The selection criterion for each metric was based on identifying where ATR (our target company) outperforms its peers. This approach effectively highlights the company’s strongest areas across the four main financial categories.

```python
metrics = ['Profitability', 'Valuation', 'Efficiency', 'Financial Strength']
bestMtr = fn.bestMetrics(df_comp, metrics, companySymbol)

```

```python
print("Best metrics for ATR:")
for group, metric in bestMtr.items():
    print(f"{group}: {metric}")
```

### Dashboard 1

```python
fig, axs = plt.subplots(2, 2, figsize=(28, 18))

color = "#4261A9FF"

fig.text(0.45, 0.96, f"Dashboard of {company[('Info', 'company_name')].iloc[0]} for ", fontsize=40, weight='bold', ha='center')
fig.text(0.74, 0.96, f"Return on Assets", fontsize=40, weight='bold', ha='center', color=color)


dim = {'title': 24, 'labels': 16, 'ticks': 14, 'text': 18}

fn.barplot_roa_by_area(ax=axs[0, 0], df=df_comp, mtr=bestMtr['Profitability'], mtr_area='Profitability', dim=dim, color=color, companySymbol=companySymbol, area='sector', top=False)
fn.barplot_roa_by_area(ax=axs[0, 1], df=df_comp, mtr=bestMtr['Profitability'], mtr_area='Profitability', dim=dim, color=color, companySymbol=companySymbol, area='industry', top=False)
fn.barplot_roa_by_area(ax=axs[1, 0], df=df_comp, mtr=bestMtr['Profitability'], mtr_area='Profitability', dim=dim, color=color, companySymbol=companySymbol, area='country', horizontal=False, addCompany=False)
fn.treemap_roa_by_area(ax=axs[1, 1], df=df_comp, mtr=bestMtr['Profitability'], mtr_area='Profitability', dim=dim, color=color, company_symbol=companySymbol)

plt.tight_layout(rect=[0, 0.03, 1, 0.9])
plt.savefig("./images/1_dashboard.png", dpi=300, bbox_inches="tight", pad_inches=0.5)
plt.show()
```

AptarGroup (ATR) shows a positive ROA of **7.13%**, which is above the average in its sector, industry, and country. However, it remains distant from the top performers, particularly in the healthcare sector where ROA values exceed 20%. Within its industry (Medical Instruments & Supplies), ATR ranks near to fifth, narrowly making the top 5. Among companies with a similar market cap, ROA values are comparable, indicating ATR performs **in line with its peers by size**.
